# 02 · Animation prototype with live force feed and logging
Tool fixed, workpiece moves at Vc (as in the sketch). `t(time)` can vary, like a₁(t) in the sketch. Forces start with a short engagement ramp; each frame is logged to CSV.
This matplotlib version is the **16-Oct prototype**. Phase 2 re-implements the same maths in a browser front-end (see PROJECT_HANDOFF.md).

```python
import sys; sys.path.insert(0,'.')
import matplotlib; matplotlib.rcParams['animation.embed_limit']=60
from IPython.display import HTML
import matplotlib.pyplot as plt
from orthocut import *; from twin_anim import run_twin
```

```python
mat  = Material('Al 7050', tau_s=303, sigma_y=505)   # sigma_y: confirm datasheet value
tool = Tool(gamma0_deg=13, mu=0.5)
step = lambda s: 0.10 if s < 3 else 0.15                # uncut thickness profile t(time) [mm]
anim, log, fig = run_twin(mat, tool, Vc=200, t_profile=step, w=2.0, duration=6, fps=20, noise=0.02, out='twin_log.csv')
plt.close(fig); HTML(anim.to_jshtml())
```

```python
log.describe().T[['mean','min','max']]
```

```python
# Save a video for the prof (needs ffmpeg)
anim.save('twin_demo.mp4', writer='ffmpeg', fps=20, dpi=110)
```
