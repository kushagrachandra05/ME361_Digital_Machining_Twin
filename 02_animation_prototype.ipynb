# 02 · Animation (scope-strict rebuild)
Pure orthogonal cutting, single shear plane, **steady state**. Inputs: Vc, t (constant), τs/σy (+ fixed γ₀, μ, w). No BUE, no thermal effects, no wear, no varying thickness/feed/properties, no start-up transient.
Consequence: forces are constant in time; the live panel logs them every frame and shows the Merchant circle from the slides.

```python
import sys; sys.path.insert(0,'.')
import matplotlib; matplotlib.rcParams['animation.embed_limit']=60
from IPython.display import HTML
import matplotlib.pyplot as plt
from orthocut import *; from twin_anim import run_twin
```

```python
# SS316L: tau_s below is a PLACEHOLDER (Mises sigma_y/sqrt3 with sigma_y ~ 290 MPa). Replace with the value you justify
# from the professor's material certificate / literature (see notebook 04 for the calibrated effective value).
sy = 290.0
mat  = Material('SS316L', tau_s=Material.tau_from_sigma_y(sy), sigma_y=sy)
tool = Tool(gamma0_deg=0.0, mu=0.5)          # gamma0=0 as in the ETH tube-cutting data; set to the actual insert later
anim, log, fig = run_twin(mat, tool, Vc=100, t=0.10, w=2.0, duration=4, fps=20, out='twin_log.csv')
plt.close(fig); HTML(anim.to_jshtml())
```
